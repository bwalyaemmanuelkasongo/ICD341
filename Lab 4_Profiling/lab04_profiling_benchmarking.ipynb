```python
import numpy as np

def square_loop(x):
    result = []
    for value in x:
        result.append(value ** 2)
    return result

def square_numpy(x):
    return x ** 2
```

```python
x = np.arange(1_000_000)

print("--- Part A: x ** 2 ---")
%timeit x ** 2

print("\n--- Part B: Python Loop ---")
%timeit square_loop(x)

print("\n--- Part B: NumPy Vectorized ---")
%timeit square_numpy(x)

print("\n--- Part C: Different Sizes ---")
%timeit np.arange(100) ** 2
%timeit np.arange(10_000) ** 2
%timeit np.arange(1_000_000) ** 2

print("\n--- Part D: Statistical Calculations ---")
data = np.random.normal(size=1_000_000)
%timeit np.mean(data)
%timeit np.std(data)
%timeit np.sum(data)
```
