```python
import numpy as np

def square_loop(x):
    result = []
    for value in x:
        result.append(value ** 2)
    return result

def square_numpy(x):
    return x ** 2
```

```python
x = np.arange(1_000_000)

print("--- Part A: x ** 2 ---")
%timeit x ** 2

print("\n--- Part B: Python Loop ---")
%timeit square_loop(x)

print("\n--- Part B: NumPy Vectorized ---")
%timeit square_numpy(x)

print("\n--- Part C: Different Sizes ---")
%timeit np.arange(100) ** 2
%timeit np.arange(10_000) ** 2
%timeit np.arange(1_000_000) ** 2

print("\n--- Part D: Statistical Calculations ---")
data = np.random.normal(size=1_000_000)
%timeit np.mean(data)
%timeit np.std(data)
%timeit np.sum(data)
```

Q2. Why might NumPy be faster?
Answer: NumPy operates in compiled C over contiguous memory, eliminating per-element Python object creation, type checks, and method dispatch. It also leverages CPU SIMD instructions.

Q3. Explain how the running time changes as the dataset becomes larger.
Answer: Running time grows roughly linearly (O(n)) with the number of elements for element-wise operations. The 100-element case is dominated by fixed overhead, while the 1,000,000-element case reveals the true per-element cost.

Q4. Which operation is fastest on your machine?
Answer: np.sum and np.mean are typically the fastest (single reduction passes). np.std is the slowest because it requires two passes (computing the mean, then summing the squared deviations).

Q5. Why should performance be measured rather than assumed?
Answer: Performance depends on hardware, library versions, memory layout, and problem size. Theoretical complexity gives scaling trends, but real-world constant factors vary. Benchmarking on the actual environment is the only reliable way to make performance decisions.
