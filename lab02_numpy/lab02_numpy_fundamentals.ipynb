```python
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print("First mark:", marks[0])
print("Last mark:", marks[-1])
print("First three marks:", marks[:3])
print("Marks from position 3 to 6:", marks[2:6])
print("Every second mark:", marks[::2])
```

```python
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Sum:", np.sum(marks))
```

### Interpretation

The mean represents the average mark obtained by the students.

The median represents the middle value when the marks are arranged in order.

The standard deviation measures how spread out the marks are.

The minimum represents the lowest mark.

The maximum represents the highest mark.

The sum represents the total of all marks.

```python
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print(data.shape)
print(data.ndim)
```

```python
print("First row:")
print(data[0])

print("Second column:")
print(data[:, 1])

print("Value 91:")
print(data[1, 1])

print("First two rows:")
print(data[:2])

print("Last two columns:")
print(data[:, 1:])
```

```python
print("Mean mark for each subject:")
print(np.mean(data, axis=0))

print("\nMean mark for each student:")
print(np.mean(data, axis=1))
```

Difference Between axis=0 and axis=1
axis=0 performs calculations down each column. In this lab, it calculates the average mark for each subject.

axis=1 performs calculations across each row. In this lab, it calculates the average mark for each student.

```python
import matplotlib.pyplot as plt

x = np.random.normal(loc=50, scale=10, size=1000)

print("Mean:", np.mean(x))
print("Median:", np.median(x))
print("Standard Deviation:", np.std(x))
print("Minimum:", np.min(x))
print("Maximum:", np.max(x))

plt.hist(x, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```
