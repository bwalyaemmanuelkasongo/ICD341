LAB 3: NumPy Vectorization and Broadcasting



Q1. What is vectorization?
Answer: Vectorization is expressing operations on whole arrays at once (like x**2 or 2*x + 5) rather than looping element by element in Python. NumPy executes these in highly optimized, compiled C code.

Q2. Why is vectorization useful?
Answer: It is significantly faster (removes Python interpreter overhead), more memory-efficient, and results in code that is easier to read and mathematically intuitive.

Q3. What is the difference between a Python loop and a NumPy vectorized operation?
Answer: A Python loop runs the interpreter once per element. A vectorized operation performs the loop inside compiled C over a contiguous memory buffer, utilizing SIMD instructions for speed.

Q4. Explain why data + adjustment works even though data is 3x3 and adjustment is 1x3.
Answer: This is broadcasting. NumPy aligns shapes from the trailing dimension. It stretches the 1x3 array down the rows (without physically copying data) so that element-wise addition can occur column by column.

Q5. Explain why subtracting a vector of means from the entire matrix is possible.
Answer: Due to broadcasting, the 1D array of means (shape (3,)) is treated as (1, 3) and stretched across the 3 rows. Subtracting it centers the data column by column.

Q6. Are the loop and vectorized results equal? Which is faster?
Answer: Yes, np.allclose() confirms the results are identical. The vectorized approach is much faster because it avoids per-element interpreter overhead and uses compiled C loops.

Q7. Explain why vectorization is particularly useful when working with large statistical datasets.
Answer: Statistical workflows repeatedly transform the entire dataset (standardizing, centering, computing residuals). On millions of observations, Python loops become a bottleneck. Vectorization removes this overhead, making interactive statistical analysis on large data feasible in Python.
