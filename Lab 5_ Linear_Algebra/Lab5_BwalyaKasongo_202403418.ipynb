```python
import numpy as np

print("--- Q1: Solve 3x + 2y = 12 and x + 4y = 10 ---")
A1 = np.array([[3, 2], [1, 4]])
b1 = np.array([12, 10])
sol1 = np.linalg.solve(A1, b1)
print(f"x = {sol1[0]:.2f}, y = {sol1[1]:.2f}")

print("\n--- Q2: 3x3 Matrix System ---")
# System:
# x + y + z = 6
# 2y + 5z = -4
# 2x + 5y - z = 27
A2 = np.array([[1, 1, 1], [0, 2, 5], [2, 5, -1]])
b2 = np.array([6, -4, 27])
sol2 = np.linalg.solve(A2, b2)
print("Solution [x, y, z]:", sol2)
```

```python
print("--- Q4: Regression using a Design Matrix ---")
X_data = np.array([30, 45, 60, 100, 150, 200, 240, 250, 300, 320])
Y_data = np.array([25, 20, 40, 65, 120, 190, 200, 200, 260, 220])

# (i) Create the design matrix (add a column of 1s for the intercept)
X_design = np.column_stack((np.ones(len(X_data)), X_data))

# Calculate coefficients using the Normal Equation: beta = (X^T X)^-1 X^T Y
beta = np.linalg.inv(X_design.T @ X_design) @ X_design.T @ Y_data
intercept, slope = beta
print(f"(i) Intercept (Beta_0): {intercept:.4f}")
print(f"    Slope (Beta_1): {slope:.4f}")

# (ii) Estimate consumption for income = 181
x_new = 181
y_pred = intercept + (slope * x_new)
print(f"\n(ii) Estimated Consumption for Income 181: {y_pred:.2f}")
```

**Q3. Explain why np.linalg.solve(A, b) is generally preferred to explicitly calculating A⁻¹b.**
*Answer:* Explicitly calculating the inverse of a matrix (A⁻¹) is computationally expensive (requires O(n³) operations) and is prone to floating-point rounding errors, especially for large or ill-conditioned matrices. `np.linalg.solve(A, b)` uses optimized algorithms (like LU decomposition) that are faster, more memory-efficient, and numerically stable.
