```python
import numpy as np
```

```python
x = np.arange(1000000)
```

```python
%timeit x ** 2
```

```python
def square_loop(x):
    result = []

    for value in x:
        result.append(value ** 2)

    return result
```

```python
%timeit square_loop(x)
```

```python
def square_numpy(x):
    return x ** 2
```

```python
%timeit square_numpy(x)
```

Why might NumPy be faster?
NumPy uses optimized compiled code and avoids the overhead of Python loops, making computations much more efficient.

```python
data = np.random.normal(size=1000000)
```

```python
%timeit np.mean(data)
```

```python
%timeit np.std(data)
```

```python
%timeit np.sum(data)
```

Which operation is fastest?
The fastest operation was np.sum(data), which took approximately 2.04 ms per loop.

Why should performance be measured rather than assumed?
Performance depends on the computer hardware, software environment, and implementation details. Benchmarking provides actual measurements instead of assumptions.
