Q1. Define a Python environment.
Answer: A Python environment is an isolated directory containing its own Python interpreter and its own set of installed packages (with specific versions). Activating it makes that interpreter and those packages the ones used by the shell/IDE, keeping projects independent and reproducible.

Q2. State two advantages of using Conda environments.
Answer:

  -Reproducibility: Exact package versions can be pinned and exported, ensuring the code runs the same way on different machines.

  -Isolation: Prevents dependency conflicts between projects and protects the base system installation.

Q3. What is JupyterLab?
Answer: JupyterLab is a web-based interactive development environment for notebooks, code, and data. It runs .ipynb notebooks (code, markdown, and output cells) against a selected kernel.

Q4. What is VS Code?
Answer: Visual Studio Code is a free, cross-platform source-code editor. With Python and Jupyter extensions, it becomes a full Python IDE with debugging, IntelliSense, linting, Git integration, and notebook support.

Q5. What is Git?
Answer: Git is a distributed version-control system that records snapshots (commits) of files over time, enabling branching, merging, reverting, and collaboration without needing a central server.

Q6. What is GitHub?
Answer: GitHub is a cloud platform for hosting Git repositories. It adds collaboration features like pull requests, issues, forks, and actions. Git is the tool; GitHub is the hosting service.

Q7. Explain the difference between git add and git commit.
Answer: git add <file> stages changes into the index (preparing them for the next commit). git commit -m "msg" records everything staged as a permanent snapshot in the repository's history.

Q8. What is the purpose of .gitignore?
Answer: It lists files and folders that Git should not track (like caches, virtual environments, secrets, or build artifacts), keeping the repository clean and preventing accidental commits of sensitive data.

Q9. Why should .ipynb_checkpoints normally be excluded from Git?
Answer: It is an auto-generated folder Jupyter creates for notebook recovery/autosave. It duplicates notebooks with random names, adds noise, causes merge conflicts, and is not part of the actual deliverable.

Q10. Explain why reproducible environments are important in statistical computing.
Answer: Statistical results depend on exact package versions and random number generators. Different versions of NumPy or SciPy can yield different numbers or defaults. Reproducible environments ensure anyone can re-run the analysis and obtain identical results, which is essential for scientific validity and peer review.

Q11. What does import numpy as np mean?
Answer: It imports the NumPy module into the current namespace and binds it to the alias np. Thereafter, np.array(...) is equivalent to numpy.array(...).

Q12. What happens if you try to import a package that isn't installed?
Answer: Python raises a ModuleNotFoundError and execution stops at that line (unless handled by a try/except block).

Q13. Explain the difference between sys.version and sys.executable.
Answer: sys.version is a string showing the Python version number (e.g., 3.12.4). sys.executable is the full file path to the Python binary currently running, confirming exactly which environment is active.

Q14. Run the notebook from VS Code and confirm that it uses the same environment as JupyterLab.
Answer: By running sys.executable in both tools, you can compare the output paths. If they both point to the icd341 environment (e.g., ...\envs\icd341\python.exe), they are using the same environment.
