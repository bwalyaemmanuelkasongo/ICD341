What is vectorization?
Vectorization is performing operations on entire arrays at once instead of using loops.

Why is vectorization useful?
It makes code simpler and faster.

Difference between loops and vectorized operations
Loops process one value at a time, while vectorized operations process entire arrays at once.

```python
marks = np.array([40, 55, 62, 70, 81, 90])

adjusted = marks + 5
percentage = marks / 100

print("Adjusted marks:")
print(adjusted)

print("\nPercentages:")
print(percentage)
```

```python
z = (marks - np.mean(marks)) / np.std(marks)

print("Z-scores:")
print(z)

print("\nMean of z:")
print(np.mean(z))
```

Verification
The mean of the standardized values (z-scores) is approximately zero, confirming that the data has been standardized correctly.

```python
data = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [90, 85, 95]
])

adjustment = np.array([5, 0, 2])

result = data + adjustment

print(result)
```

Why does broadcasting work?
Broadcasting allows NumPy to automatically expand the smaller array to match the shape of the larger array.

data has shape (3,3)

adjustment has shape (3,)

NumPy applies the adjustment values to each row automatically.

```python
data = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

means = data.mean(axis=0)

print("Column means:")
print(means)

centered = data - means

print("\nCentered data:")
print(centered)
```

```python
print(centered.mean(axis=0))
```

Why is subtracting the vector means from the entire matrix possible?
The means array contains one mean value for each column.

NumPy broadcasting automatically applies each mean to the corresponding column of the matrix.

This allows all columns to be centered without using loops.

```python
x = np.arange(1000000)

result_loop = []

for value in x:
    result_loop.append(value**2 + 2*value + 1)

result_vectorized = x**2 + 2*x + 1

print(np.allclose(result_loop, result_vectorized))
```

## Are the two results equal?

Yes, both approaches produce the same result.

## Which approach is easier to read?

The NumPy vectorized approach is easier to read.

## Which approach is faster?

The NumPy vectorized approach is faster.

## Why?

NumPy performs computations using optimized compiled code and avoids Python loop overhead.

```python
np.random.seed(123)

x = np.random.normal(100, 15, 100000)

z = (x - x.mean()) / x.std()

print("Mean of z:", z.mean())
print("Standard deviation of z:", z.std())
```

Why is vectorization useful for large statistical datasets?
Vectorization allows NumPy to process entire arrays at once instead of one value at a time.

This makes computations significantly faster, reduces code complexity, and improves performance when working with large datasets.

```python
import numpy as np

def standardize(x):
    return (x - np.mean(x)) / np.std(x)
```

```python
a = np.array([10, 20, 30, 40, 50])
b = np.array([5, 10, 15, 20, 25])
c = np.array([100, 120, 140, 160, 180])

print("Array A:")
print(standardize(a))

print("\nArray B:")
print(standardize(b))

print("\nArray C:")
print(standardize(c))
```

```python
def standardize_columns(x):
    return (x - np.mean(x, axis=0)) / np.std(x, axis=0)
```

```python
data = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

result = standardize_columns(data)

print("Original Data:")
print(data)

print("\nStandardized Data:")
print(result)
```

```python
print("Column Means:")
print(result.mean(axis=0))

print("\nColumn Standard Deviations:")
print(result.std(axis=0))
```
