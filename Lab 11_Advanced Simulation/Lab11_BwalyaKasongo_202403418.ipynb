```python
import numpy as np
import matplotlib.pyplot as plt

print("--- Q1: Inverse Transform Sampling (Exponential, rate=3) ---")
# The CDF of exponential is F(x) = 1 - exp(-rate * x)
# The inverse is F_inv(u) = -ln(1 - u) / rate
n_samples = 5000
rate = 3
u = np.random.uniform(0, 1, n_samples)
exp_samples = -np.log(1 - u) / rate
print(f"Mean of samples: {np.mean(exp_samples):.4f} (Expected: {1/rate:.4f})")

print("\n--- Q2: Sample from Categorical Distribution ---")
probs = [0.1, 0.6, 0.3]
categories = [0, 1, 2]
cat_samples = np.random.choice(categories, size=10000, p=probs)
print(f"Sample proportions: {np.bincount(cat_samples) / len(cat_samples)}")

print("\n--- Q3: Rejection Sampling for Beta(2,2) ---")
# Target: Beta(2,2) -> PDF is proportional to x * (1-x) for x in [0,1]. Max is at x=0.5 (value = 0.25)
# Proposal: Uniform(0,1) -> PDF = 1. 
# We need a constant M such that M * q(x) >= p(x). M = 0.25 is sufficient.
def beta_pdf(x):
    return x * (1 - x) # Ignoring the normalizing constant (1/6)

samples = []
M = 0.25
while len(samples) < 5000:
    x_prop = np.random.uniform(0, 1)
    u = np.random.uniform(0, 1)
    if u <= beta_pdf(x_prop) / M:
        samples.append(x_prop)
print(f"Mean of Rejection Samples: {np.mean(samples):.4f} (Expected: 0.5)")
```

**Q1. Generate 5,000 exponential samples using inverse transform with rate 3.**
*Answer:* Using the inverse CDF method, we set \(F^{-1}(u) = -\frac{\ln(1-u)}{\lambda}\) where \(\lambda=3\). By generating 5,000 uniform random numbers and applying this formula, we obtained the exponential samples. The sample mean was approximately 0.333, matching the theoretical mean of \(1/3\).

**Q2. Sample from a categorical distribution with probabilities [0.1, 0.6, 0.3].**
*Answer:* We used `np.random.choice` with the specified probabilities. For 10,000 samples, the empirical proportions were approximately [0.1, 0.6, 0.3], confirming the sampling works correctly.

**Q3. Implement rejection sampling for Beta(2,2).**
*Answer:* The target Beta(2,2) PDF is proportional to \(x(1-x)\). We used a Uniform(0,1) proposal distribution and a constant \(M=0.25\) (the maximum value of the target PDF). Samples were accepted if a uniform random number \(u \le \frac{f(x)}{M}\). The mean of the accepted samples was approximately 0.5, which matches the theoretical mean of Beta(2,2).

**Q4. Describe the four main particle-filter steps.**
*Answer:* 
1. **Initialization:** Generate an initial set of particles (samples) to represent the prior distribution.
2. **Prediction (Propagation):** Move each particle forward in time according to the state transition model (adding process noise).
3. **Update (Weighting):** Calculate the likelihood of the observed data given each particle's state, and assign weights accordingly.
4. **Resampling:** Discard particles with low weights and duplicate particles with high weights to focus computational effort on the most probable states.

**Q5. Explain what a non-uniform SBC rank histogram may indicate.**
*Answer:* Simulation-Based Calibration (SBC) checks if a Bayesian model is well-calibrated. If the rank histogram is **non-uniform** (e.g., U-shaped, bell-shaped, or skewed), it indicates a problem with the model. A U-shape means the posterior is too narrow (overconfident), while a bell shape means the posterior is too wide (underconfident). A uniform histogram indicates the model is perfectly calibrated.
