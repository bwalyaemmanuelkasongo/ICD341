```python
import numpy as np

print("--- Q1: Gradient Descent for f(x) = (x - 10)^2 ---")
# The derivative of f(x) is f'(x) = 2(x - 10)
def df(x):
    return 2 * (x - 10)

x_current = 0.0  # Starting point
learning_rate = 0.1
iterations = 20

for i in range(iterations):
    gradient = df(x_current)
    x_current = x_current - learning_rate * gradient

print(f"Optimized x after {iterations} iterations: {x_current:.4f} (Theoretical min is x = 10)")
```

```python
from scipy.optimize import minimize

print("--- Q2: BFGS vs CG on f(x,y) = (x-4)^2 + (y-6)^2 ---")
def objective(vars):
    x, y = vars
    return (x - 4)**2 + (y - 6)**2

initial_guess = [0, 0]

# BFGS Method
res_bfgs = minimize(objective, initial_guess, method='BFGS')
print(f"BFGS Result: x={res_bfgs.x[0]:.4f}, y={res_bfgs.x[1]:.4f} (Iterations: {res_bfgs.nit})")

# Conjugate Gradient (CG) Method
res_cg = minimize(objective, initial_guess, method='CG')
print(f"CG Result:   x={res_cg.x[0]:.4f}, y={res_cg.x[1]:.4f} (Iterations: {res_cg.nit})")
```

```python
from scipy import stats, optimize
import numpy as np # Just in case it needs to be re-imported

print("--- Q4: MLE for Normal Mean using SciPy ---")
np.random.seed(42)
data = np.random.normal(loc=50, scale=5, size=100) 

def neg_log_likelihood(mu, data):
    return -np.sum(stats.norm.logpdf(data, loc=mu, scale=5))

res_mle = optimize.minimize(neg_log_likelihood, x0=[0], args=(data,))
print(f"MLE Estimated Mean: {res_mle.x[0]:.4f}")
print(f"Actual Sample Mean: {np.mean(data):.4f}")

print("\n--- Q5: Sigmoid Function ---")
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print(f"Sigmoid(0): {sigmoid(0):.4f}")
print(f"Sigmoid(-2): {sigmoid(-2):.4f}")
```

```python
print("--- Q6: Logistic Regression on Student Data ---")
X_log = np.array([1, 2, 3, 4, 5, 6, 7, 8])
Y_log = np.array([0, 0, 0, 0, 1, 1, 1, 1])

X_design_log = np.column_stack((np.ones(len(X_log)), X_log))

def neg_log_likelihood_logit(beta, X, Y):
    p = sigmoid(X @ beta)
    epsilon = 1e-15
    p = np.clip(p, epsilon, 1 - epsilon)
    return -np.sum(Y * np.log(p) + (1 - Y) * np.log(1 - p))

init_beta = np.zeros(X_design_log.shape[1])
res_logit = optimize.minimize(neg_log_likelihood_logit, init_beta, args=(X_design_log, Y_log))
print(f"Fitted Intercept: {res_logit.x[0]:.4f}")
print(f"Fitted Slope: {res_logit.x[1]:.4f}")
```

*Answer:* Both algorithms successfully found the minimum at x = 4 and y = 6. BFGS is a quasi-Newton method that builds an approximation of the Hessian matrix and typically converges in fewer iterations compared to the Conjugate Gradient (CG) method, which is often better suited for much larger problems where computing the Hessian is infeasible.

**Q3. Explain the effect of changing the learning rate.**
*Answer:* The learning rate controls the step size during gradient descent. 
- If it is **too small**, the algorithm takes tiny steps and takes a very long time to converge.
- If it is **too large**, the algorithm may overshoot the minimum and bounce back and forth, failing to converge (diverging).
- An **optimal** learning rate allows the algorithm to reach the minimum efficiently.
