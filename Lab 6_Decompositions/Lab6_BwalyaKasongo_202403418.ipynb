```python
import numpy as np

print("--- Q1: Covariance Matrix & Cholesky Decomposition ---")
np.random.seed(42)
# Create a positive-definite matrix
M = np.random.rand(3, 3)
cov_matrix = M @ M.T
print("Covariance Matrix:\n", np.round(cov_matrix, 3))

# Perform Cholesky decomposition
L = np.linalg.cholesky(cov_matrix)
print("\nCholesky Lower Triangular (L):\n", np.round(L, 3))

print("\n--- Q2: QR Decomposition ---")
A = np.array([[1, 2], [3, 4], [5, 6]])
Q, R = np.linalg.qr(A)
print("Q Matrix (Orthogonal):\n", np.round(Q, 3))
print("R Matrix (Upper Triangular):\n", np.round(R, 3))

print("\n--- Q3: SVD Reconstruction ---")
U, S, Vt = np.linalg.svd(A)
# Reconstruct the original matrix A = U * S * Vt
S_matrix = np.zeros((A.shape[0], A.shape[1]))
np.fill_diagonal(S_matrix, S)
A_reconstructed = U @ S_matrix @ Vt
print("Original A:\n", A)
print("Reconstructed A:\n", np.round(A_reconstructed, 3))

print("\n--- Q4: Condition Number ---")
cond_num = np.linalg.cond(A)
print(f"Condition Number of A: {cond_num:.3f}")
```
