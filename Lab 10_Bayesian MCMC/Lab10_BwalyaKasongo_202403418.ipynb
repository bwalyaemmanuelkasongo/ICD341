```python
import numpy as np

print("--- Q1: Random-Walk MH for Standard Normal Target ---")
np.random.seed(42)

def target_pdf(x):
    # Standard normal PDF (ignoring constants)
    return np.exp(-0.5 * x**2)

n_samples = 10000
samples = np.zeros(n_samples)
current = 0.0  # Starting point
proposal_std = 1.0  # Step size for random walk

for i in range(n_samples):
    proposal = np.random.normal(current, proposal_std)
    acceptance_ratio = target_pdf(proposal) / target_pdf(current)
    if np.random.rand() < acceptance_ratio:
        current = proposal
    samples[i] = current

print(f"Mean of MH samples: {np.mean(samples):.4f} (Target is 0)")
print(f"Std Dev of MH samples: {np.std(samples):.4f} (Target is 1)")
```

```python
import numpy as np

print("--- Q2: Bernoulli Probability with Beta(2,2) Prior ---")
# Data: 8 successes in 12 trials
successes = 8
trials = 12
failures = trials - successes

# Prior is Beta(2,2). 
# For a Bernoulli likelihood, the posterior is conjugate: Beta(alpha + successes, beta + failures)
alpha_post = 2 + successes
beta_post = 2 + failures

# The posterior mean is the estimated probability
posterior_mean = alpha_post / (alpha_post + beta_post)
print(f"Posterior Parameters: Alpha = {alpha_post}, Beta = {beta_post}")
print(f"Estimated Probability of Success (Posterior Mean): {posterior_mean:.4f}")

# Generate 10000 samples from the posterior to find the 95% credible interval
posterior_samples = np.random.beta(alpha_post, beta_post, 10000)
credible_interval = np.percentile(posterior_samples, [2.5, 97.5])
print(f"95% Credible Interval: {credible_interval}")
```

**Q3. Explain trace plots, ACF, R-hat, and ESS.**
*Answer:* 
- **Trace plot:** A line plot of the sampled values over iterations. It is used to visually check if the Markov chain has converged and is exploring the target distribution (it should look like a "hairy caterpillar").
- **ACF (Autocorrelation Function):** Measures the correlation between samples at different lags. In MCMC, we want autocorrelation to drop to zero quickly so that samples are roughly independent.
- **R-hat (Gelman-Rubin statistic):** Compares the variance between multiple chains to the variance within each chain. An R-hat close to 1.0 (usually < 1.05) indicates convergence.
- **ESS (Effective Sample Size):** Estimates the number of independent samples the chain is equivalent to. Because MCMC samples are autocorrelated, the ESS is always less than the total number of draws.
