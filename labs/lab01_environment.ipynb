QUESTIONS

1. What is a Python environment?
A separate, self-contained space with its own Python and its own installed packages, so it doesn't affect anything else on your computer.

2. Why create a separate environment for a course?
It keeps the course's packages and versions from clashing with other projects, and you can delete it easily when you're done.

3. What does conda activate icd341 do?
It switches your terminal into the icd341 environment, so any Python or packages you run come from that environment.

4. What is the purpose of python=3.12?
It tells conda to install Python version 3.12 in the environment, so everything runs on the version the course expects.
