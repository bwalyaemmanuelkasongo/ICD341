```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data from your data folder
df = pd.read_csv("../data/maize_production.csv")

# Create a simple plot
plt.plot(df['Year'], df['Production'], marker='o', color='green')
plt.title("Maize Production Over Time")
plt.xlabel("Year")
plt.ylabel("Production (Tonnes)")

# Save the graph directly into the 'outputs' folder!
plt.savefig("../outputs/graphs.png")
plt.show()
```
