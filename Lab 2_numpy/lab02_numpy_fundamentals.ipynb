```python
import numpy as np

# Part A - Creating arrays
x = np.array([10, 20, 30, 40, 50])
print("Array:", x)
print("Shape:", x.shape)
print("Dimensions:", x.ndim)
print("Data type:", x.dtype)
```

```python
# Part B - Array creation methods
a = np.zeros(10)
b = np.ones(10)
c = np.arange(1, 11)
d = np.linspace(0, 1, 11)

print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)
```

```python
# Part C - Indexing and slicing
marks = np.array([45, 67, 82, 56, 91, 73, 64, 88])

print("1. First mark:", marks[0])
print("2. Last mark:", marks[-1])
print("3. First three marks:", marks[:3])
print("4. Marks from position 3 to 6:", marks[2:6])
print("5. Every second mark:", marks[::2])
```

```python
# Part D - Statistical calculations
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Sum:", np.sum(marks))
```

```python
# Part E - Two-dimensional arrays
data = np.array([
    [45, 67, 82],
    [56, 91, 73],
    [64, 88, 76]
])

print("Shape:", data.shape)
print("Dimensions:", data.ndim)
print("\nTask 4 Extractions:")
print("1. First row:", data[0])
print("2. Second column:", data[:, 1])
print("3. Value 91:", data[1, 1])
print("4. First two rows:\n", data[:2])
print("5. Last two columns:\n", data[:, -2:])
```

```python
# Part F - Statistical application
print("Subject means (axis=0):", np.mean(data, axis=0))
print("Student means (axis=1):", np.mean(data, axis=1))
```

```python
# Challenge - Simulated data and histogram
import matplotlib.pyplot as plt

x = np.random.normal(loc=50, scale=10, size=1000)

print("Mean:", np.mean(x))
print("Median:", np.median(x))
print("Std Dev:", np.std(x))
print("Min:", np.min(x))
print("Max:", np.max(x))

plt.hist(x, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Distribution of Simulated Data")
plt.show()
```
