```python
import numpy as np

# Data from the question
data = np.array([3, 4, 5, 7, 8, 9])
B = 10000  # Number of bootstrap samples
np.random.seed(42) # For reproducibility

print("--- Q1: Bootstrap the Mean & 95% Percentile Interval ---")
# Generate 10,000 bootstrap samples of the mean
boot_means = np.array([np.mean(np.random.choice(data, size=len(data), replace=True)) for _ in range(B)])
# Calculate the 95% percentile interval (2.5th and 97.5th percentiles)
mean_ci = np.percentile(boot_means, [2.5, 97.5])
print(f"Bootstrap Mean: {np.mean(boot_means):.4f}")
print(f"95% Percentile Interval for Mean: {mean_ci}")

print("\n--- Q2: Bootstrap the Median ---")
# Generate 10,000 bootstrap samples of the median
boot_medians = np.array([np.median(np.random.choice(data, size=len(data), replace=True)) for _ in range(B)])
median_ci = np.percentile(boot_medians, [2.5, 97.5])
print(f"Bootstrap Median: {np.median(boot_medians):.4f}")
print(f"95% Percentile Interval for Median: {median_ci}")
```

```python
print("--- Q3: Parametric Bootstrap from a Fitted Normal Model ---")
# 1. Fit a normal distribution to the original data
mu_hat = np.mean(data)
sigma_hat = np.std(data, ddof=1)

# 2. Generate parametric bootstrap samples
param_boot_means = np.array([
    np.mean(np.random.normal(loc=mu_hat, scale=sigma_hat, size=len(data))) 
    for _ in range(B)
])

param_ci = np.percentile(param_boot_means, [2.5, 97.5])
print(f"Fitted Normal Parameters: Mean = {mu_hat:.4f}, Std Dev = {sigma_hat:.4f}")
print(f"Parametric Bootstrap 95% CI for Mean: {param_ci}")
```

**Q4. Explain one way data leakage can occur during model validation.**
*Answer:* Data leakage occurs when information from outside the training dataset is used to create the model. A common way this happens during validation is **preprocessing on the entire dataset before splitting**. For example, if you standardize (scale) the data or impute missing values using the mean of the *entire* dataset before doing a train-test split, the test set's information leaks into the training process. Preprocessing steps must only be fitted on the training data.
