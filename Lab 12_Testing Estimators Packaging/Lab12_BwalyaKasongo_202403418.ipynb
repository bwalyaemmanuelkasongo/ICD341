```python
import numpy as np

print("--- Q1: Median Function with Tests ---")
def calculate_median(data):
    if len(data) == 0:
        raise ValueError("Data cannot be empty")
    return np.median(data)

# Test 1: Known result
assert calculate_median([1, 2, 3, 4, 5]) == 3.0
assert calculate_median([3, 1, 2]) == 2.0
print("Q1 Known Result Tests Passed!")

# Test 2: Empty input
try:
    calculate_median([])
except ValueError as e:
    print(f"Q1 Empty Input Test Passed! (Caught error: {e})")


print("\n--- Q2: Test Probability Estimate [0,1] ---")
def probability_estimate(data):
    return np.mean(data) 

# Tests
assert 0 <= probability_estimate([0, 1, 1, 0]) <= 1
assert 0 <= probability_estimate([1, 1, 1, 1]) <= 1
print("Q2 Tests Passed! Estimate is always in [0,1].")


print("\n--- Q3: Custom Transformer (Standard Deviation Scaler) ---")
class StdScaler:
    def fit(self, X, y=None):
        self.std_ = np.std(X, axis=0)
        self.std_[self.std_ == 0] = 1.0 
        return self
        
    def transform(self, X):
        return X / self.std_

    def fit_transform(self, X, y=None):
        return self.fit(X).transform(X)

# Test the transformer
X_dummy = np.array([[1, 10], [2, 20], [3, 30]])
scaler = StdScaler()
X_scaled = scaler.fit_transform(X_dummy)
print("Q3 Original Data:\n", X_dummy)
print("Q3 Scaled Data:\n", X_scaled)


print("\n--- Q4: Create pyproject.toml ---")
pyproject_content = """
[build-system]
requires = ["setuptools>=61.0"]
build-backend = "setuptools.build_meta"

[project]
name = "icd341_lab12"
version = "0.1.0"
description = "A test package for Lab 12"
"""
with open("pyproject.toml", "w") as f:
    f.write(pyproject_content)
print("Q4 pyproject.toml created successfully in the current folder!")
```
