```python
import numpy as np

np.random.seed(42) # For reproducibility

print("--- Q1: 10000 N(50, 10^2) Observations ---")
norm_data = np.random.normal(loc=50, scale=10, size=10000)
print(f"Sample Mean: {np.mean(norm_data):.2f}, Sample Std: {np.std(norm_data):.2f}")

print("\n--- Q2: 10000 Bernoulli(p=0.3) ---")
bern_data = np.random.binomial(n=1, p=0.3, size=10000)
print(f"Sample Proportion of Successes: {np.mean(bern_data):.4f} (Expected: 0.3)")

print("\n--- Q3: Monte Carlo Estimate of Integral of x^3 from 0 to 1 ---")
# Generate uniform points [0,1], cube them, and average
x_unif = np.random.uniform(0, 1, 1000000)
integral_estimate = np.mean(x_unif**3)
print(f"Estimated Integral: {integral_estimate:.4f} (True value is 1/4 = 0.25)")
```

```python
print("\n--- Q4: Estimate Pi with 1,000,000 points ---")
points = np.random.uniform(-1, 1, size=(1000000, 2))
# Check which points fall inside the unit circle (x^2 + y^2 <= 1)
inside_circle = np.sum(np.sum(points**2, axis=1) <= 1)
# Area of circle / Area of square = Pi / 4
pi_estimate = (inside_circle / 1000000) * 4
print(f"Estimated Pi: {pi_estimate:.5f}")

print("\n--- Q5: Estimate P(X > 1.96) for N(0,1) ---")
z_scores = np.random.normal(0, 1, 1000000)
prob_estimate = np.mean(z_scores > 1.96)
print(f"Estimated Probability: {prob_estimate:.4f} (True value ~ 0.0250)")
```

```python
from scipy import stats

print("\n--- Q6: Importance Sampling for E(X^2) where X ~ U(0,1) ---")
# Target distribution: Uniform(0,1). True E(X^2) = 1/3 ≈ 0.3333
# We will use a Beta(2,2) distribution as our proposal distribution q(x)
n_samples = 100000
samples_q = np.random.beta(2, 2, n_samples) # samples from proposal q
pdf_p = stats.uniform.pdf(samples_q, 0, 1)  # pdf of target p
pdf_q = stats.beta.pdf(samples_q, 2, 2)    # pdf of proposal q

# Importance sampling estimator: E[X^2 * p(x)/q(x)]
weights = pdf_p / pdf_q
is_estimate = np.mean((samples_q**2) * weights)
print(f"Importance Sampling Estimate: {is_estimate:.4f}")
```

**Q7. Explain the difference between importance sampling and control variates.**
*Answer:* 
- **Importance Sampling** changes the probability distribution from which we draw samples. Instead of sampling from the target distribution, we sample from a "proposal" distribution that focuses on the important regions (rare events or peaks) and then re-weight the samples to correct for the difference.
- **Control Variates** keeps the original sampling distribution but introduces a second variable (the control variate) that is highly correlated with the variable of interest but whose true expected value is known. We use this known value to adjust the estimate and reduce variance.
